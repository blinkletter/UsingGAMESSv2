### The GAMESS Folder

Within the `gamess` directory are several **important files**. Two of which you may chose to edit. Below are the files and some brief descriptions.

| File                  | Description                                                    |
| :-------------------- | :------------------------------------------------------------- |
| auxdata               | A directory with important data for calculations               |
| *gamess.00.x*         | The *GAMESS* application binary (the old 2020 binary was named *gamess.30Jun2020R1.x*) |
| *gms-files.csh*       | A shell script that defines environmental variables for *GAMESS*|

| tests                 | A directory of sample jobs to test your installation           |

The *rungms* and *gms* scripts contain variables that can be changed to **customize** the behaviour of *GAMESS* on your computer.

### Editing RUNGMS

Open *rungms* in a **text editor**. The first lines of the script define directories for the **scratch files**. These are important to define. The default is the home directory of the *rungms* script, but you will likely want your result files to appear somewhere else. For my first installation I changed the first four lines of the script…

```text
set TARGET=sockets
set SCR=./
set USERSCR=./
set GMSPATH=./
```
to…
```text
set TARGET=sockets
set SCR=/Applications/gamess/scr
set USERSCR=/Users/Barry/Documents/CompChem/GAMESS-SCR
set GMSPATH=/Applications/gamess
```

This means that the *GAMESS* scratch directory (`SCR`) is no longer the home directory of the application but a directory within it called *scr*. The scratch directory contains huge **temporary** files that are generally **deleted** when the calculation is **complete**. If it crashes, the files may **remain**. I wanted a separate directory so that any leftover junk would be easy to delete every now and then. Be sure to create the `scr` directory.

The user scratch directory (`USERSCR`) contains data files that are written out with the completion of a calculation. These files contain **orbital and frequency data** among many other sets of useful information. We only occasionally need these files, but they will now be stored in a directory of my choice rather than the `gamess` directory. Be sure to create that directory where it is expected to be.

The location of the *GAMESS* binary file is specified with `GMSPATH`.

#### Two extra lines for the notebooks

The notebooks in this book run *rungms* directly and want the data files (`.dat`, `.trj`, …) to land **next to the log file** of each job rather than in one shared folder. In my 2023 build the same block of *rungms* (near line 100) now reads:

```text
set TARGET=sockets
set SCR=~/gamess/scr
set USERSCR=~/gamess/restart
if ($?GMS_USERSCR) set USERSCR=$GMS_USERSCR   # the notebooks set this to the job's folder
set GMSPATH=~/gamess
set DEL_OLD=yes
```

- The `if ($?GMS_USERSCR)` line lets *Python* choose `USERSCR` for each job. Without it, the data files go to the `USERSCR` folder named in *rungms*. Everything still works, you just have to look for them there.
- `DEL_OLD=yes` deletes data files left over from an earlier run of a job with the **same name**. Without it, *rungms* refuses to start and prints `Please save, rename, or erase these files from a previous run`.

Make a backup copy of *rungms* before you edit it, for example `cp rungms rungms.orig`.

## Running GAMESS

### From the terminal

*GAMESS* can be run from the **terminal command line** by calling *gms*. For example, the command `gms cyclohexane.inp` will call the *rungms* script using 1 processor and produce a result file called `cyclohexane.log`.

`gms -l result.log -n 2 cyclohexane.inp` will do the same, but write the result to a file named `result.log` and use 2 processors in your computer. (The *gms* script I use is described on the [compiling page](1B-CompileInstruction2023.md).)

### From a notebook

In this book we will instead run *GAMESS* **from *Python***. The book comes with two small files that sit next to the notebooks:

| File | What it does | Do you edit it? |
| :--- | :--- | :--- |
| `gamess_settings.py` | Says where *GAMESS* lives on **your** computer | **Yes, once** |
| `gamess_tools.py` | Functions to run a job and read the log file | No, but do read it |

Open `gamess_settings.py` in your text editor and change `GAMESS_DIR` to the folder that holds `rungms` and `gamess.00.x`. Save it and restart the notebook kernel. That is the only place where your *GAMESS* location is written down.

The function `run_gamess()` in `gamess_tools.py` does the same job as the *gms* script: it deletes leftover scratch files for the job, calls `rungms` with `subprocess`, writes everything *GAMESS* prints into a `.log` file and tells you whether the job finished normally. All jobs from the notebooks write their results into a folder called `runs/`.

### Settings

Run the cell below. It loads the helper module and shows where it thinks *GAMESS* is. If the paths are wrong, edit `gamess_settings.py`, then restart the kernel and run the cell again.

````python
# ===== SETTINGS =========================================================
# Your GAMESS location is set ONCE, in gamess_settings.py (next to this
# notebook). Edit that file, save it, and restart the kernel.
from pathlib import Path
import gamess_tools as gt

print("GAMESS folder:", gt.gamess_dir())
print("Run script:   ", gt.run_script())
# ========================================================================
````

### Check the installation

`check_installation()` looks for the pieces that *GAMESS* needs. On a Mac it also checks the **shared-memory limit**. *MacOS* resets this limit to a few megabytes every time the computer restarts, and then every *GAMESS* job fails. The fix is two `sudo sysctl` commands that you type in the terminal after each restart (see step 8 of the [compiling instructions](1C-CompilingGAMESS-USonM5MacBookAir.md)).

````python
gt.check_installation()
````

### Your first job

The book includes a small input file, `inputs/water_test.inp`. It optimizes the structure of water at a low level of theory and takes a few seconds. We will learn what each line means in the next chapter. For now we simply read it with *Python* and print it.

````python
test_input = Path("inputs/water_test.inp")
print(test_input.read_text())
````

Now send it to *GAMESS*. This one line replaces typing `gms water_test.inp` in the terminal.

````python
log = gt.run_gamess(test_input)
log
````

A successful job prints `finished normally`. If it says `FAILED`, open the `.log` file and read the **first** error message near the end of the file. The most common causes are a wrong `GAMESS_DIR`, a missing `scr` folder and the shared-memory limit.

The log file is long. Let us look at its **last few lines**, which is what you would do with `tail` in the terminal.

````python
lines = log.read_text().splitlines()
print(f"{log.name} has {len(lines)} lines. The last 12 are:\n")
print("\n".join(lines[-12:]))
````

*GAMESS* found the lowest-energy structure in a few steps. The helper function `nserch_energies()` pulls out the energy (in hartree) at each step of the optimization. How it does this is the subject of the [next page](1A_Unix.ipynb).

````python
for step, energy in gt.nserch_energies(log):
    print(f"step {step}:  E = {energy:.10f} hartree")
````

### Run a *GAMESS* test job (optional)

*GAMESS* comes with a set of test inputs in its `tests/standard` folder. Each input lists the answer it should give in its comment lines. `exam01` optimizes the structure of methylene, CH<sub>2</sub>, and should finish with `FINAL E= -37.2380397698`. We can run it straight from the *GAMESS* folder and compare.

````python
exam01 = gt.gamess_dir() / "tests" / "standard" / "exam01.inp"
log01 = gt.run_gamess(exam01)

expected = -37.2380397698
found = gt.final_energy(log01)
print(f"expected {expected:.10f}, found {found:.10f}, difference {found - expected:.1e} hartree")
````

If the two energies agree to about 10<sup>-9</sup> hartree, your *GAMESS* is working. You can run the full test suite from the terminal with `runall` and `checktst` as described on the compiling pages.

## Installing *GamessQ*

*GamessQ* is a batch controller for *GAMESS* provided by the Gordon group. It is available from the [same web site](https://www.msg.chem.iastate.edu/gamess/). It can **queue** a whole batch of files and send them to be executed one at a time while you sleep. In this book a *Python* loop in a notebook can do the same job, but *GamessQ* is still handy when you build molecules in *MacMolPlt* and just want to drop the input files somewhere.

### Obtain and Install

Download the compressed file and extract it. Put the resulting folder in `/Applications`. In the `/Applications/GamessQ` directory we will see the following files&hellip;

| File                  | Description                                                    |
| :-------------------- | :------------------------------------------------------------- |
| GamessQ.app           | The application for the MacOS GUI interface                    |
| gamessq.html          | Instructions                                                   |
| gamessqd              | The background daemon application                              |

### Running *GamessQ*

You will need to run the *gamessqd* application **first**. If you double-click on it, it will run via a terminal window. The *Terminal* app will open and run the executable binary *gamessqd*. It will keep running as long as that terminal window is open. After this **daemon** has been launched, you will double-click on the *GamessQ* application to **open** the graphic interface. *GamessQ* will not run if *gamessqd* has not already been activated.

**Click** on the *GamessQ* icon to run the program. In the menus, open `GamessQ` &rarr; `Preferences` and you will see two **important settings**. The first is the location of the **spool directory** for *GamessQ*. This is where *GamessQ* will move the final result files from a *GAMESS* calculation. The second is the location of the directory for the ***rungms* script** that runs *GAMESS*. Create a directory for the result files and point *GamessQ* there. And then enter the location of the `gamess` folder. Now it is **ready to launch** jobs.

### Using *GamessQ*

To use *GamessQ*, you **drag and drop** the text input files onto its window. This will place the job in the **queue**. You can see when jobs are **waiting, running and finished**. You can drop in dozens of jobs and let them run over the weekend this way.

There is a **command line version** of *GamessQ* that you can compile and run directly on your computer. This can be useful if you want to set up a dedicated computer to run *GAMESS*. You will be able to *telnet* (look it up) using *ssh* (look it up) into your account on that computer. Then you can transfer over input files and submit jobs to the queue using *gamessq* directly.
